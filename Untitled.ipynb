```python
import time
import json
from confluent_kafka import Consumer, KafkaException, OFFSET_BEGINNING
import base64 # (Not used if storing path, but useful if you did base64 encoding earlier)
import os
from cassandra.cluster import Cluster, ResultSet # Import Cassandra classes
from cassandra.auth import PlainTextAuthProvider # Might be needed for authentication, though not for default Docker
from cassandra.query import prepared_statement_cache # Best practice for prepared statements

# --- Consumer Configuration ---
MAX_FETCH_SIZE_BYTES = 6 * 1024 * 1024 # 6 MB

consumer_config = {
    'bootstrap.servers': 'localhost:9092',
    'group.id': 'image_processing_group',
    'auto.offset.reset': 'earliest',
    'enable.auto.commit': True,
    'auto.commit.interval.ms': 5000,
    'client.id': 'image-consumer-app',
    'max.partition.fetch.bytes': MAX_FETCH_SIZE_BYTES,
    'fetch.max.bytes': 52428800, # Max bytes to fetch per consumer session poll call
    'broker.version.fallback': '1.0.0',
    'api.version.request': 'true'
}

# --- Cassandra Configuration ---
CASSANDRA_CONTACT_POINTS = ['localhost'] # Replace with your Cassandra container name if not using localhost
CASSANDRA_PORT = 9042
CASSANDRA_KEYSPACE = 'football_sensor_data_keyspace' # Use a distinct keyspace name

# --- Global Cassandra Session (initialized once) ---
cluster = None
session = None
insert_statement = None # Will be a prepared statement

# --- Kafka Callback for Partition Assignment (Existing) ---
def reset_offset(consumer, partitions):
    """Callback to reset offsets for assigned partitions."""
    for p in partitions:
        p.offset = OFFSET_BEGINNING # Set offset to beginning for each assigned partition
    consumer.assign(partitions)
    print(f"INFO: Assigned partitions with offset reset: {partitions}")

# --- Main Consumer Logic ---
def run_kafka_to_cassandra_consumer():
    global cluster, session, insert_statement # Use global to access and modify

    consumer = None # Initialize consumer here
    try:
        # Initialize Kafka Consumer
        consumer = Consumer(consumer_config)
        consumer.subscribe([TOPIC_NAME], on_assign=reset_offset)
        print(f"--- Starting Consumer for Topic: {TOPIC_NAME} in Group: {consumer_config['group.id']} ---")

        # --- Cassandra Connection Setup ---
        # Connect to Cassandra
        print(f"INFO: Connecting to Cassandra at {CASSANDRA_CONTACT_POINTS}:{CASSANDRA_PORT}...")
        cluster = Cluster(CASSANDRA_CONTACT_POINTS, port=CASSANDRA_PORT)
        session = cluster.connect()
        print("INFO: Connected to Cassandra cluster.")

        # Create Keyspace if it doesn't exist
        # Replication factor 1 for single-node Docker setup
        session.execute(f"""
            CREATE KEYSPACE IF NOT EXISTS {CASSANDRA_KEYSPACE}
            WITH replication = {{'class': 'SimpleStrategy', 'replication_factor': 1}};
        """)
        print(f"INFO: Keyspace '{CASSANDRA_KEYSPACE}' ensured.")
        session.set_keyspace(CASSANDRA_KEYSPACE) # Set the active keyspace

        # Create Table if it doesn't exist
        # sensor_id: Partition key (ensures all data for one sensor goes to one partition)
        # timestamp_ms: Clustering key (orders data by time within a sensor's partition)
        # image_storage_path: Will store the path/URL to the large binary data
        # data_size_bytes: Useful for knowing the original size of the large data
        session.execute("""
            CREATE TABLE IF NOT EXISTS sensor_image_data (
                sensor_id text,
                timestamp_ms bigint,
                data_type text,
                sequence_num int,
                source_app text,
                data_size_bytes bigint,
                image_storage_path text, -- Will store path to simulated S3/disk location
                PRIMARY KEY ((sensor_id), timestamp_ms)
            ) WITH CLUSTERING ORDER BY (timestamp_ms DESC);
        """)
        print("INFO: Cassandra table 'sensor_image_data' ensured.")

        # Prepare the INSERT statement once for efficiency
        insert_statement = session.prepare("""
            INSERT INTO sensor_image_data (sensor_id, timestamp_ms, data_type, sequence_num, source_app, data_size_bytes, image_storage_path)
            VALUES (?, ?, ?, ?, ?, ?, ?)
        """)
        print("INFO: Cassandra INSERT statement prepared.")


        # --- Kafka Message Consumption Loop ---
        while True:
            msg = consumer.poll(timeout=1.0) # Poll for 1 second
            if msg is None:
                # print("DEBUG: No message received in poll.") # Uncomment for debugging
                continue

            if msg.error():
                if msg.error().code() == KafkaException._PARTITION_EOF:
                    # End of partition event (not an error, just an info)
                    # print(f"INFO: End of partition reached {msg.topic()} [{msg.partition()}]")
                    pass
                elif msg.error():
                    # Real error, e.g., network issues
                    print(f"ERROR: Kafka Error: {msg.error()}")
                    # Consider breaking or more robust error handling
                continue

            # --- Process Kafka Message ---
            key_bytes = msg.key()
            value_bytes = msg.value() # This is your 5MB binary data

            if not key_bytes or not value_bytes:
                print(f"WARN: Skipping message with missing key or value. Topic: {msg.topic()}, Offset: {msg.offset()}")
                continue

            try:
                metadata = json.loads(key_bytes.decode('utf-8'))
                print(f"INFO: Received message from topic '{msg.topic()}', partition [{msg.partition()}], offset {msg.offset()}. "
                      f"Key: {metadata.get('sensor_id')}, Value size: {len(value_bytes) / (1024 * 1024):.2f} MB.")

                # --- Handle Large Binary Data (Value) ---
                # In a real scenario, 'value_bytes' would be uploaded to S3, GCS, Azure Blob Storage here.
                # For this practice, we'll just simulate a path.
                # A unique path could be: {sensor_id}/{timestamp_ms}.bin or {topic}/{partition}/{offset}.bin
                simulated_storage_path = (
                    f"simulated_s3_bucket/football_data/{metadata.get('sensor_id', 'unknown')}/"
                    f"{metadata.get('timestamp_ms', int(time.time()*1000))}_seq{metadata.get('sequence_num', 'N/A')}.bin"
                )
                
                # You'd write value_bytes to this path:
                # with open(f"/tmp/{os.path.basename(simulated_storage_path)}", "wb") as f:
                #     f.write(value_bytes)
                # print(f"DEBUG: Saved {len(value_bytes)} bytes to {simulated_storage_path}")

                # --- Insert Data into Cassandra ---
                session.execute(
                    insert_statement,
                    (metadata.get('sensor_id'),
                     metadata.get('timestamp_ms'),
                     metadata.get('data_type'),
                     metadata.get('sequence_num'),
                     metadata.get('source_app'),
                     len(value_bytes), # Storing original size
                     simulated_storage_path) # Storing the *path/reference*
                )
                print(f"INFO: Successfully inserted metadata for sensor '{metadata.get('sensor_id')}' (path: '{simulated_storage_path}') into Cassandra.")

            except json.JSONDecodeError as e:
                print(f"ERROR: Failed to decode JSON key for message at offset {msg.offset()}: {e}. Key: {key_bytes.decode('utf-8', errors='ignore')}")
            except Exception as e:
                print(f"ERROR: Failed to process or insert message into Cassandra at offset {msg.offset()}: {e}")
                # In production, consider a Dead Letter Queue (DLQ) for failed messages
                
    except KafkaException as e:
        print(f"CRITICAL ERROR: Kafka Consumer Exception: {e}")
    except Exception as e:
        print(f"CRITICAL ERROR: Consumer setup or loop failed: {e}")
    finally:
        if consumer:
            print("INFO: Closing Kafka consumer.")
            consumer.close() # Always close the consumer cleanly
        if cluster:
            print("INFO: Shutting down Cassandra cluster connection.")
            cluster.shutdown() # Always shutdown Cassandra cluster connection cleanly
        print("--- Consumer Finished ---")

# --- Main Execution Block ---
if __name__ == "__main__":
    run_kafka_to_cassandra_consumer()
```
